```python
#PinakRokde-23070521196
import numpy as np
import matplotlib.pyplot as plt

x=np.array([1,2,3,4,5])
y=np.array([20,40,50,65,80])

print("X =", x)
print("Y =", y)

x_mean=np.mean(x)
y_mean=np.mean(y)

print("X Mean =", x_mean)
print("Y Mean =", y_mean)

numerator=np.sum((x-x_mean)*(y-y_mean))
denominator=np.sum((x-x_mean)**2)
m=numerator/denominator
print("Slope =", m)
c=y_mean-(m*x_mean)
print("Y-intercept =", c)

y_pred=m*x+c
print("Predicted Y =", y_pred)
for i in range(len(x)):
  print("Actual Y =", y[i], "Predicted Y =", y_pred[i])

residuals=y-y_pred
print("Residuals =", residuals)
for i in range(len(x)):
  print("Actual Y =", y[i], "Residual =", residuals[i])

mae= np.mean(np.abs(residuals))
print("Mean Absolute Error =", mae)

mse=np.mean(residuals**2)
print("Mean Squared Error =", mse)

rmse=np.sqrt(mse)
print("Root Mean Squared Error =", rmse)

plt.scatter(x, y,
            color='green',
            s=80,
            label='Actual Data')

plt.plot(x,
         y_pred,
         color='pink',
         linewidth=2,
         label='Regression Line')

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Linear Regression")

plt.grid(True)
plt.legend()

plt.show()
```
